## 一、热身：定义简单函数

三个函数分别练习：无参数函数、单参数函数、带循环的多参数函数。
`goat_pad` 在字符串前后各拼上一个 `"goat"`，`print_n` 用 `for` 循环把同一个字符串打印 n 次，每次一行。

```python
def say_hi():
    """不接收参数，打印 Hello, world!。"""
    print("Hello, world!")


def goat_pad(s):
    """把字符串 s 前后各加上 goat 后打印。"""
    print("goat" + s + "goat")


def print_n(s, n):
    """把字符串 s 打印 n 次，每次一行。"""
    for _ in range(n):
        print(s)
```

```python
say_hi()
goat_pad("hello")
goat_pad("Python")
print_n("Hello, world!", 3)
```

## 二、欧几里得算法

欧几里得算法基于恒等式 gcd(m, n) = gcd(n, m % n)：不断用余数替换较大的数，直到余数为 0，
此时的除数就是最大公约数。用 `while` 循环实现即可，无需递归，也不需要做输入检查。

```python
def gcd(m, n):
    """返回两个整数的最大公约数（欧几里得算法）。"""
    while n != 0:
        m, n = n, m % n
    return m
```

```python
pairs = [(2018, 2019), (1200, 300), (5040, 60)]
for m, n in pairs:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")
```

### 关于负数参数

Python 的 `%` 运算结果的符号跟随除数，因此算法仍然会终止，但返回值可能带负号：
`gcd(-4, 6)` 返回 `2`，而 `gcd(4, -6)` 返回 `-2`，`gcd(-4, -6)` 返回 `-2`。
最大公约数按定义应当是非负整数，所以这个行为对负数输入并不完全有意义；
如果需要严谨处理，可在函数开头对两个参数取绝对值。

```python
print("gcd(-4, 6)  =", gcd(-4, 6))
print("gcd(4, -6)  =", gcd(4, -6))
print("gcd(-4, -6) =", gcd(-4, -6))
```

## 三、近似欧拉数 e

自然常数 e 可以写成无穷级数：

e = Σ(k=0→∞) 1/k! = 1 + 1/1! + 1/2! + 1/3! + …

也可以写成极限形式：e = lim(x→∞) (1 + 1/x)^x。

四个函数分别对应：取 x = n 的极限近似、前 n 项级数求和、在误差 ε 内逼近 e、以及打印求和表。

```python
import math


def euler_limit(n):
    """用 (1 + 1/n)^n 近似 e。"""
    return (1 + 1 / n) ** n


def euler_infinite_sum(n):
    """返回级数前 n 项之和（n 从 1 开始计数），n = 0 时返回 0。"""
    total = 0.0
    factorial = 1
    for k in range(n):          # k = 0, 1, ..., n-1，正好 n 项
        total += 1 / factorial
        factorial *= (k + 1)
    return total


def euler_approx(epsilon):
    """累加级数，直到结果与 math.exp(1) 的差小于 epsilon。"""
    target = math.exp(1)
    total = 0.0
    factorial = 1
    k = 0
    while abs(total - target) >= epsilon:
        total += 1 / factorial
        k += 1
        factorial *= k
    return total


def print_euler_sum_table(n):
    """打印 k = 1 到 n 的 euler_infinite_sum(k)。"""
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))
```

```python
print("euler_limit(10)     =", euler_limit(10))
print("euler_limit(100)    =", euler_limit(100))
print("euler_limit(10000)  =", euler_limit(10000))
print()
print("euler_infinite_sum(0)  =", euler_infinite_sum(0))
print("euler_infinite_sum(1)  =", euler_infinite_sum(1))
print("euler_infinite_sum(4)  =", euler_infinite_sum(4))
print("euler_infinite_sum(20) =", euler_infinite_sum(20))
print("math.exp(1)            =", math.exp(1))
print()
print("euler_approx(1e-2) =", euler_approx(1e-2))
print("euler_approx(1e-6) =", euler_approx(1e-6))
print("euler_approx(1e-12)=", euler_approx(1e-12))
```

```python
print_euler_sum_table(10)
```

## 四、测试整数属性

正整数 n 是 2 的幂，当且仅当 n = 2^p（p ≥ 0）。不使用 `math.sqrt`，只用除法和取模：
反复用 2 整除 n，若最后得到 1，则 n 是 2 的幂；否则不是。

`is_power(b, n)` 把上面的 2 换成任意正整数 b；注意 b = 1 时 1 的任何次幂都是 1，
循环会无法终止，因此需要单独处理。

```python
def is_power_of_2(n):
    """判断正整数 n 是否为 2 的幂。"""
    while n % 2 == 0 and n > 1:
        n //= 2
    return n == 1


def is_power(b, n):
    """判断正整数 n 是否为 b 的幂（b, n 均为正整数）。"""
    if b == 1:
        return n == 1
    while n % b == 0 and n > 1:
        n //= b
    return n == 1
```

```python
print("is_power_of_2:")
for n in [1, 2, 3, 4, 6, 8, 16, 18, 64, 100]:
    print(f"  is_power_of_2({n}) = {is_power_of_2(n)}")

print()
print("is_power:")
for b, n in [(3, 9), (3, 27), (3, 28), (2, 8), (2, 12), (5, 1), (10, 1000), (4, 16)]:
    print(f"  is_power({b}, {n}) = {is_power(b, n)}")
```
