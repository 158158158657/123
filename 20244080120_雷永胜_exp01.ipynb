### 实验2：函数、条件、循环与递归

`score_level` 用多分支 `if` 完成分段，并对越界输入抛 `ValueError`；随后练习 for / while 两种循环、
递归与迭代两种阶乘实现，以及可变参数和元组返回。

```python
def score_level(score):
    """按分数返回等级；超出 0~100 时抛出 ValueError。"""
    if isinstance(score, bool) or not isinstance(score, (int, float)):
        raise ValueError(f"成绩必须是数值: {score!r}")
    if score < 0 or score > 100:
        raise ValueError(f"成绩必须在 0~100 之间: {score}")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"
```

```python
for score in [100, 90, 89.5, 80, 79, 60, 59.9, 0]:
    print(f"score_level({score}) = {score_level(score)}")

print()
for bad in [-1, 101, 120]:
    try:
        score_level(bad)
    except ValueError as exc:
        print("ValueError:", exc)
```

```python
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for   结果:", total_for)
print("while 结果:", total_while)
print("两者一致  :", total_for == total_while == sum(range(1, 101)))
```

```python
def factorial_recursive(n):
    """递归版阶乘，约定 0! = 1。"""
    if n < 0:
        raise ValueError("n 必须是非负整数")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    """迭代版阶乘，约定 0! = 1。"""
    if n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result
```

```python
import math

for n in [0, 1, 5, 10]:
    r = factorial_recursive(n)
    it = factorial_iterative(n)
    print(f"n={n:<2} 递归={r:<10} 迭代={it:<10} math.factorial={math.factorial(n):<10} 一致={r == it == math.factorial(n)}")
```

```python
def summary(values):
    """接收非空数值列表，返回 (最小值, 最大值, 平均值) 元组。"""
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)


scores = [85, 58, 92, 76, 60, 89]
low, high, average = summary(scores)
print(f"最小值={low}  最大值={high}  平均值={average:.2f}")
```

```python
def sum_all(*args):
    """返回所有参数之和。"""
    return sum(args)


print("sum_all()             =", sum_all())
print("sum_all(5)            =", sum_all(5))
print("sum_all(1, 2, 3)      =", sum_all(1, 2, 3))
print("sum_all(1.5, 2.5, -1) =", sum_all(1.5, 2.5, -1))
```

### 实验4：文件读写、异常处理与模块

在 Notebook 同级目录创建 `exp01_data` 目录并写入 `students.txt`。文件里既有合法记录，也有三类非法记录，
分别用于验证字段数量、成绩格式、成绩范围和空姓名等校验逻辑。

```python
import os

print("当前工作目录:", os.getcwd())
data_dir = os.path.join(".", "exp01_data")
os.makedirs(data_dir, exist_ok=True)
txt_path = os.path.join(data_dir, "students.txt")
print("数据目录    :", data_dir)
print("数据文件    :", txt_path)
```

```python
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(txt_path, "w", encoding="utf-8") as handle:
    handle.write(content)

with open(txt_path, encoding="utf-8") as handle:
    lines = handle.read().splitlines()

print("总行数（含表头）:", len(lines))
for line in lines[1:]:
    print("  ", line.split(","))
```

```python
def parse_student(line):
    """把一行 'name,score' 解析成字典，非法记录抛出 ValueError。"""
    parts = [part.strip() for part in line.split(",")]
    if len(parts) != 2:
        raise ValueError(f"字段数量错误（应为 2 个，实际 {len(parts)} 个）")
    name, score_text = parts
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩不是整数: {score_text!r}") from None
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围: {score}")
    return {"name": name, "score": score}
```

```python
valid_records = []
invalid_records = []

with open(txt_path, encoding="utf-8") as handle:
    next(handle)                                  # 跳过表头
    for lineno, line in enumerate(handle, start=2):
        line = line.strip()
        if not line:
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as exc:
            invalid_records.append((lineno, line, str(exc)))

print("有效记录:")
for record in valid_records:
    print("  ", record)
print("无效记录:")
for lineno, line, reason in invalid_records:
    print(f"   第 {lineno} 行 {line!r}: {reason}")
```

```python
valid_count = len(valid_records)
invalid_count = len(invalid_records)
valid_scores = [record["score"] for record in valid_records]
avg_score = sum(valid_scores) / valid_count
pass_rate = sum(1 for score in valid_scores if score >= 60) / valid_count

report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as handle:
    handle.write(f"有效记录数: {valid_count}\n")
    handle.write(f"无效记录数: {invalid_count}\n")
    handle.write(f"平均分: {avg_score:.2f}\n")
    handle.write(f"及格率: {pass_rate:.1%}\n")

print("w 模式写入后的 report.txt:")
print(open(report_path, encoding="utf-8").read())

with open(report_path, "a", encoding="utf-8") as handle:
    handle.write("—— 本行由 a 模式追加，原有内容保留 ——\n")

print("a 模式追加后的 report.txt:")
print(open(report_path, encoding="utf-8").read())
```

**说明：** `w` 模式会先清空文件再写入，因此重复运行只会保留本次内容；`a` 模式把新内容追加到末尾，
原有内容保持不变。

```python
import pickle

pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as handle:
    pickle.dump(valid_records, handle)

with open(pkl_path, "rb") as handle:
    restored = pickle.load(handle)

print("恢复的数据:", restored)
print("与原数据相等:", restored == valid_records)
```

### 实验5：类、继承与对象复制

`Student` 封装姓名和成绩，`is_passed()` 判断是否及格，`__str__()` 提供可读输出；
`GraduateStudent` 通过 `super().__init__()` 复用父类初始化，并重写 `__str__()`；
最后比较直接赋值、`copy.copy()` 与 `copy.deepcopy()` 在处理嵌套可变对象时的区别。

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name!r}, score={self.score}) -> {status}"
```

```python
students = [Student(record["name"], record["score"]) for record in valid_records]
for student in students:
    print(student)
print()
print("及格人数:", sum(1 for student in students if student.is_passed()))
```

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"{super().__str__()}, research_topic={self.research_topic!r}"


grad = GraduateStudent("Alice", 85, "数据挖掘")
print(grad)
print("是否及格:", grad.is_passed())
print("是 Student 的实例:", isinstance(grad, Student))
```

```python
import copy


def fresh_object():
    return {"name": "Alice", "scores": [85, 90]}


original = fresh_object()
assigned = original
assigned["scores"].append(100)
print("直接赋值  -> original =", original)

original = fresh_object()
shallow = copy.copy(original)
shallow["scores"].append(100)
print("浅拷贝    -> original =", original, "（内部列表被共享）")

original = fresh_object()
deep = copy.deepcopy(original)
deep["scores"].append(100)
print("深拷贝    -> original =", original, "（内部列表独立）")
```

**说明：** 直接赋值只是新增一个指向同一对象的引用；`copy.copy()` 复制外层对象，但内部的列表仍然共享；
`copy.deepcopy()` 递归复制所有嵌套对象，因此修改副本不会影响原对象。

## 三、实验小结

1. Python 的 `/` 总是产生浮点数，`//` 是向下取整除法，`%` 与 `//` 配合可做整数分解。
2. 字符串不可变，任何“修改”都要通过切片、拼接或 `join()` 构造新对象。
3. 循环与递归可以互相替代；递归要注意终止条件 `n == 0`。
4. 文件读写要显式指定 `encoding="utf-8"`，异常处理让程序在遇到坏数据时仍能继续处理后面的记录。
5. `is` 比较对象身份，`==` 比较值；浅拷贝与深拷贝的差异只在对象内部还有可变对象时才显现。
